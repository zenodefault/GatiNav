# Synthetic fixture validation

Plot ground truth and, when the EKF is available, its output.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))
from python.tests.conftest import synthetic_circle

fixture = synthetic_circle()
gt = fixture.gt_pose['position_enu']
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.figure(figsize=(6, 6))
plt.plot(gt[:, 0], gt[:, 1], label='ground truth')
try:
    from python.ekf.ekf import ErrorStateEKF
except ModuleNotFoundError:
    ErrorStateEKF = None

if ErrorStateEKF is not None:
    ekf = ErrorStateEKF()
    estimate = []
    for i in range(len(fixture.t)):
        ekf.predict(fixture.gyro[i], fixture.accel[i], 1 / 100)
        estimate.append(ekf.position.copy())
    estimate = np.asarray(estimate)
    plt.plot(estimate[:, 0], estimate[:, 1], label='EKF output')

plt.xlabel('East (m)')
plt.ylabel('North (m)')
plt.axis('equal')
plt.legend()
plt.show()
```
